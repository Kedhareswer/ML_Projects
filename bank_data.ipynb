```python
# Cell: 1 - Imports and global config
import os
import gc
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import roc_auc_score, log_loss
from sklearn.linear_model import LogisticRegression
import lightgbm as lgb
from catboost import CatBoostClassifier, Pool
from sklearn.ensemble import StackingClassifier
import shap

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
sns.set(style='whitegrid', font_scale=1.1)
```

```python
# Cell: 2 - Paths & Load data
DATA_DIR = '.'  # change if files are elsewhere
train_path = os.path.join(DATA_DIR, '/kaggle/input/playground-series-s5e8/train.csv')
test_path = os.path.join(DATA_DIR, '/kaggle/input/playground-series-s5e8/test.csv')
sample_sub_path = os.path.join(DATA_DIR, '/kaggle/input/playground-series-s5e8/sample_submission.csv')

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)
sample_submission = pd.read_csv(sample_sub_path)

print("Train shape:", train.shape)
print("Test shape:", test.shape)
train.head(3)
```

#### 2.2 Numerical / categorical split and basic stats


```python
# Cell: 5 - infer dtypes
def infer_feature_types(df, cat_threshold=15):
    numeric = df.select_dtypes(include=[np.number]).columns.tolist()
    # exclude target from categoricals if present
    categorical = [c for c in df.columns if c not in numeric and c != 'y']
    # for numeric columns with few unique values, treat as categorical
    for c in numeric:
        if c != 'y' and train[c].nunique() <= cat_threshold:
            categorical.append(c)
            numeric.remove(c)
    return numeric, categorical

num_cols, cat_cols = infer_feature_types(train.drop(columns=['y']))
print("Numerical:", num_cols)
print("Categorical:", cat_cols)
train[num_cols].describe().T
```

#### 3. Preprocessing & Feature Engineering
- Handle missing values (simple strategy)
- Build ColumnTransformer: scale numerics, encode categoricals
- Keep pipeline generic so tournament judges can inspect easily


```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Copy original data
train_copy = train.copy()
test_copy = test.copy()

# Ensure categorical columns are strings
for c in cat_cols:
    train_copy[c] = train_copy[c].astype(str)
    test_copy[c] = test_copy[c].astype(str)

# Fill missing values
for c in num_cols:
    med = train_copy[c].median()
    train_copy[c] = train_copy[c].fillna(med)
    test_copy[c] = test_copy[c].fillna(med)

for c in cat_cols:
    train_copy[c] = train_copy[c].fillna('missing')
    test_copy[c] = test_copy[c].fillna('missing')

# ColumnTransformer with updated OneHotEncoder syntax
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
    ],
    remainder='drop'
)

# Fit-prep to get transformed feature names (optional)
combined = pd.concat([train_copy.drop(columns=['y']), test_copy], axis=0)
preprocessor.fit(combined)
```

### 4. Modeling Strategy
- Use **StratifiedKFold** (5 folds) to estimate performance (ROC AUC and LogLoss)
- Train Logistic Regression baseline, LightGBM, CatBoost
- Generate OOF (out-of-fold) predictions for stacking


```python
# Cell: 9 - CV and utility functions
NFOLDS = 5
folds = StratifiedKFold(n_splits=NFOLDS, shuffle=True, random_state=RANDOM_STATE)

def cv_oof_predict(clf, X, y, X_test, name='model'):
    """
    Returns OOF predictions (train) and averaged test predictions (test).
    clf: sklearn-compatible estimator or Pipeline
    X: DataFrame (train features)
    y: Series (target)
    X_test: DataFrame (test features)
    """
    oof_train = np.zeros(len(X))
    oof_test = np.zeros(len(X_test))
    oof_test_skf = np.zeros((NFOLDS, len(X_test)))
    
    for i, (tr_idx, val_idx) in enumerate(folds.split(X, y)):
        print(f"{name} - Fold {i+1}/{NFOLDS}")
        X_tr, X_val = X.iloc[tr_idx], X.iloc[val_idx]
        y_tr, y_val = y.iloc[tr_idx], y.iloc[val_idx]
        
        clf.fit(X_tr, y_tr)
        oof_train[val_idx] = clf.predict_proba(X_val)[:, 1]
        oof_test_skf[i, :] = clf.predict_proba(X_test)[:, 1]
        
        score_auc = roc_auc_score(y_val, oof_train[val_idx])
        score_ll = log_loss(y_val, oof_train[val_idx])
        print(f"  Fold {i+1} AUC: {score_auc:.5f}, LogLoss: {score_ll:.5f}")
    
    oof_test[:] = oof_test_skf.mean(axis=0)
    cv_auc = roc_auc_score(y, oof_train)
    cv_ll = log_loss(y, oof_train)
    print(f"{name} CV AUC: {cv_auc:.5f}, CV LogLoss: {cv_ll:.5f}")
    return oof_train, oof_test
```

#### 5. Baseline: Logistic Regression

```python
# Cell: 10 - baseline logistic using pipeline
X = train_copy.drop(columns=['y']).reset_index(drop=True)
y = train_copy['y'].reset_index(drop=True)
X_test = test_copy.reset_index(drop=True)

pipe_lr = Pipeline([
    ('pre', preprocessor),
    ('clf', LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))
])

oof_lr, test_preds_lr = cv_oof_predict(pipe_lr, X, y, X_test, name='LogisticRegression')
```

### 6. LightGBM (tabular champion)

```python
# Transform features
X_trans = preprocessor.transform(X)
X_test_trans = preprocessor.transform(X_test)
print("Transformed shapes:", X_trans.shape, X_test_trans.shape)

def lgb_train_cv(X_arr, y_arr, X_test_arr, params=None):
    if params is None:
        params = {
            'objective': 'binary',
            'metric': 'auc',
            'verbosity': -1,
            'boosting_type': 'gbdt',
            'seed': RANDOM_STATE,
            'learning_rate': 0.05,
            'num_leaves': 31,
            'n_estimators': 1000,
            'early_stopping_rounds': 100
        }
    oof = np.zeros(len(X_arr))
    test_preds = np.zeros(len(X_test_arr))
    for i, (tr_idx, val_idx) in enumerate(folds.split(X_arr, y_arr)):
        print(f"LGBM - fold {i+1}")
        X_tr, X_val = X_arr[tr_idx], X_arr[val_idx]
        y_tr, y_val = y_arr[tr_idx], y_arr[val_idx]
        clf = lgb.LGBMClassifier(**params)
        clf.fit(X_tr, y_tr, eval_set=[(X_val, y_val)], eval_metric='auc')
        oof[val_idx] = clf.predict_proba(X_val)[:, 1]
        test_preds += clf.predict_proba(X_test_arr)[:, 1] / NFOLDS
        fold_auc = roc_auc_score(y_val, oof[val_idx])
        fold_ll = log_loss(y_val, oof[val_idx])
        print(f"  Fold {i+1} AUC: {fold_auc:.5f}, LogLoss: {fold_ll:.5f}")
    print("LGBM CV AUC: ", roc_auc_score(y_arr, oof))
    print("LGBM CV LogLoss: ", log_loss(y_arr, oof))
    return oof, test_preds

# Run training
oof_lgb, test_preds_lgb = lgb_train_cv(X_trans, y.values, X_test_trans)
```

### 7. CatBoost (handles categoricals natively — optional)
- For CatBoost we will pass categorical indices. We'll create a version with raw DataFrame inputs.


```python
# Cell: 12 - CatBoost CV using DataFrame input and providing cat_feature indices
# Get cat indices relative to full X (train_copy)
cat_feature_indices = [i for i, c in enumerate(train_copy.drop(columns=['y']).columns) if c in cat_cols]
print("Cat feature indices:", cat_feature_indices)

def catboost_cv(X_df, y_arr, X_test_df, cat_features_idx, params=None):
    if params is None:
        params = {
            'iterations': 1000,
            'learning_rate': 0.05,
            'depth': 6,
            'loss_function': 'Logloss',
            'eval_metric': 'AUC',
            'random_seed': RANDOM_STATE,
            'verbose': 0,
            'early_stopping_rounds': 100
        }
    oof = np.zeros(len(X_df))
    test_preds = np.zeros(len(X_test_df))
    for i, (tr_idx, val_idx) in enumerate(folds.split(X_df, y_arr)):
        print(f"CatBoost - fold {i+1}")
        X_tr, X_val = X_df.iloc[tr_idx], X_df.iloc[val_idx]
        y_tr, y_val = y_arr[tr_idx], y_arr[val_idx]
        train_pool = Pool(X_tr, y_tr, cat_features=cat_features_idx)
        val_pool = Pool(X_val, y_val, cat_features=cat_features_idx)
        clf = CatBoostClassifier(**params)
        clf.fit(train_pool, eval_set=val_pool, use_best_model=True)
        oof[val_idx] = clf.predict_proba(X_val)[:,1]
        test_preds += clf.predict_proba(X_test_df)[:,1] / NFOLDS
        print(f"  Fold {i+1} AUC: {roc_auc_score(y_val, oof[val_idx]):.5f}")
    print("CatBoost CV AUC:", roc_auc_score(y_arr, oof))
    print("CatBoost CV LogLoss:", log_loss(y_arr, oof))
    return oof, test_preds

# Prepare raw df with same filled NA treatment
X_df = train_copy.drop(columns=['y'])
X_test_df = test_copy.copy()
oof_cat, test_preds_cat = catboost_cv(X_df, y.values, X_test_df, cat_feature_indices)
```

### 8. Compare model CV results

```python
# Cell: 13 - compare AUC and LogLoss for OOFs
from math import isfinite

def print_scores(name, oof):
    auc = roc_auc_score(y, oof)
    ll = log_loss(y, oof)
    print(f"{name}: AUC = {auc:.5f}, LogLoss = {ll:.5f}")

print_scores('LogisticRegression', oof_lr)
print_scores('LightGBM', oof_lgb)
print_scores('CatBoost', oof_cat)
```

### 9. Feature Importance (LightGBM)

```python
# Cell: 14 - train a single lgb model on full data for feature importance (simple)
# Warning: uses transformed features; to map back to names we reconstruct feature names from preprocessor
feature_names = []
# numeric names
feature_names.extend(num_cols)
# cat names from fitted OneHotEncoder
ohe = preprocessor.named_transformers_['cat']
if hasattr(ohe, 'get_feature_names_out'):
    cat_ohe_names = list(ohe.get_feature_names_out(cat_cols))
else:
    # older sklearn
    cat_ohe_names = []
feature_names.extend(cat_ohe_names)

# Train a final LightGBM on full training set (transformed)
clf_full = lgb.LGBMClassifier(objective='binary', learning_rate=0.05, n_estimators=500, random_state=RANDOM_STATE)
clf_full.fit(X_trans, y)
imp = clf_full.feature_importances_
# get top 30 features
imp_df = pd.DataFrame({'feature': feature_names, 'importance': imp}).sort_values('importance', ascending=False).head(30)
plt.figure(figsize=(8,10))
sns.barplot(x='importance', y='feature', data=imp_df)
plt.title('LightGBM feature importances (top 30)')
plt.show()
```

#### 10. Stacking Ensemble
- Use OOF predictions from base models as meta features.
- Final meta-model: Logistic Regression (simple, robust)

```python
# Cell: 15 - Create stacked features (OOF)
stack_train = pd.DataFrame({
    'lr': oof_lr,
    'lgb': oof_lgb,
    'cat': oof_cat
})
stack_test = pd.DataFrame({
    'lr': test_preds_lr,
    'lgb': test_preds_lgb,
    'cat': test_preds_cat
})

print("Stack shapes:", stack_train.shape, stack_test.shape)

# Meta-model CV
meta = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)
oof_stack, test_stack = cv_oof_predict(meta, stack_train, y, stack_test, name='StackMeta')
print("Stack meta CV AUC:", roc_auc_score(y, oof_stack))
print("Stack meta CV LogLoss:", log_loss(y, oof_stack))
```

#### 11. Final training on full train set and submission
- We'll pick the stacking pipeline (meta) for final predictions.
- Retrain base models on full data and produce test predictions consistently.


```python
# Cell: 16 - Train base models on full data then meta
# 1) retrain Logistic (pipeline)
pipe_lr.fit(X, y)
final_test_lr = pipe_lr.predict_proba(X_test)[:,1]

# 2) retrain LGB on transformed full
clf_lgb_full = lgb.LGBMClassifier(**{
    'objective':'binary','learning_rate':0.05,'n_estimators':500,'random_state':RANDOM_STATE
})
clf_lgb_full.fit(X_trans, y)
final_test_lgb = clf_lgb_full.predict_proba(X_test_trans)[:,1]

# 3) retrain CatBoost on full
train_pool_full = Pool(X_df, y, cat_features=cat_feature_indices)
clf_cat_full = CatBoostClassifier(iterations=1000, learning_rate=0.05, depth=6, random_seed=RANDOM_STATE, verbose=0)
clf_cat_full.fit(train_pool_full, use_best_model=False)
final_test_cat = clf_cat_full.predict_proba(X_test_df)[:,1]

# meta stack features
final_stack_test = pd.DataFrame({
    'lr': final_test_lr,
    'lgb': final_test_lgb,
    'cat': final_test_cat
})

# final meta model (train on full stack_train)
meta_final = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)
meta_final.fit(stack_train, y)
final_test_preds = meta_final.predict_proba(final_stack_test)[:,1]

# Save submission
submission = sample_submission.copy()
# assume sample submission has 'id' and 'y' or similar; match column name for y
if 'y' in submission.columns:
    submission['y'] = final_test_preds
else:
    # find last column, set predictions
    submission.iloc[:, -1] = final_test_preds

submission.to_csv('submission.csv', index=False)
print("Submission saved to submission.csv")
```
